## Bagian A - Implementasi Recipe Audit Station CSV

Recipe ini digunakan untuk melakukan quality control (QC) data stasiun observasi dengan memeriksa struktur data, missing value, duplikasi timestamp, dan format waktu.

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Timestamp':['2025-01-01 00:00','2025-01-01 01:00','2025-01-01 01:00',np.nan,'2025-01-01 03:00'],
    'Temperature':[25,26,26,27,28],
    'Humidity':[80,78,78,75,74]
}

df = pd.DataFrame(data)
df
```

### Memeriksa Struktur Data

```python
df.info()
```

### Memeriksa Missing Value

```python
print(df.isnull().sum())
```

### Memeriksa Duplikasi Timestamp

```python
duplicate_timestamp = df['Timestamp'].duplicated()
print('Jumlah timestamp duplikat:', duplicate_timestamp.sum())
```

### Konversi Format Waktu

```python
df['Timestamp'] = pd.to_datetime(df['Timestamp'], errors='coerce')
df.head()
```

### Membuat Laporan Audit

```python
audit_report = {
'Jumlah Data': len(df),
'Missing Value': df.isnull().sum().sum(),
'Duplicate Timestamp': df['Timestamp'].duplicated().sum()
}

audit_report
```

### Kesimpulan Recipe Audit

- Terdapat missing value pada kolom Timestamp.
- Terdapat timestamp yang duplikat.
- Format waktu berhasil dikonversi menjadi datetime.

## Recipe Perhitungan Porositas dari Density Log

Persamaan:

φ = (ρma - ρb)/(ρma - ρf)

```python
import matplotlib.pyplot as plt

rho_ma = 2.65
rho_f = 1.0
rho_b = [2.30, 2.35, 2.40, 2.45, 2.50]
```

```python
porosity = []

for density in rho_b:
    phi = (rho_ma-density)/(rho_ma-rho_f)
    porosity.append(round(phi*100,2))

porosity
```

```python
result = pd.DataFrame({
'Bulk Density (g/cc)': rho_b,
'Porosity (%)': porosity
})

result
```

### Visualisasi Hasil

```python
plt.figure(figsize=(6,4))
plt.plot(porosity, rho_b, marker='o')
plt.gca().invert_yaxis()
plt.xlabel('Porosity (%)')
plt.ylabel('Bulk Density (g/cc)')
plt.title('Porosity from Density Log')
plt.grid()
plt.show()
```

### Interpretasi

- Nilai porositas menurun ketika densitas batuan meningkat.
- Densitas tinggi mengindikasikan ruang pori yang lebih kecil.
- Metode density porosity umum digunakan dalam evaluasi reservoir.
